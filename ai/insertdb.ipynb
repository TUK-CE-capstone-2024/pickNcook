```python
import pandas as pd
import mysql.connector

# 엑셀 파일 불러오기
file_path = r'C:\TB_RECIPE_SEARCH(1~1001).xlsx'
df = pd.read_excel(file_path)

# 열 이름을 DB 테이블에 맞게 변경 (필요 시 수동 매핑)
df.columns = [
    'RCP_TTL', 'CKG_NM', 'CKG_MTH_ACTO_NM', 'CKG_STA_ACTO_NM',
    'CKG_MTRL_ACTO_NM', 'CKG_KND_ACTO_NM', 'CKG_MTRL_CN', 'CKG_INBUN_NM',
    'CKG_DODF_NM', 'CKG_TIME_NM', 'RCP_IMG_URL'
]

# MySQL 연결 설정
conn = mysql.connector.connect(
    host="127.0.0.1",
    port=3306,
    user="root",
    password="1234",
    database="pickandcookdb"
)
cursor = conn.cursor()

# INSERT 쿼리 작성
insert_query = """
    INSERT INTO RECIPE (
        RCP_TTL, CKG_NM, CKG_MTH_ACTO_NM, CKG_STA_ACTO_NM,
        CKG_MTRL_ACTO_NM, CKG_KND_ACTO_NM, CKG_MTRL_CN, CKG_INBUN_NM,
        CKG_DODF_NM, CKG_TIME_NM, RCP_IMG_URL
    ) VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
"""

# 한 행씩 INSERT 수행 (NaN 처리 포함)
for index, row in df.iterrows():
    values = [None if pd.isna(x) else x for x in row]
    cursor.execute(insert_query, tuple(values))


# 커밋 및 연결 종료
conn.commit()
cursor.close()
conn.close()

print("데이터 삽입 완료!")
```
